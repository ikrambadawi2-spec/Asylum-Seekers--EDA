# Data Understanding

##  Column Descriptions

| # | Column Name | Description | Data Type | Example / Notes |
|---|---|---|---|---|
| 1 | `Year` | The year in which the asylum-seeker statistics were recorded. | Integer | 2000, 2001, 2020, etc. |
| 2 | `Country / territory of asylum/residence` | The country or territory where the asylum seeker submitted their claim or resides while seeking asylum. | Categorical | Egypt, Germany, France, etc. |
| 3 | `Origin` | The country or territory of origin/nationality of the asylum seeker. | Categorical | Sudan, Syria, Afghanistan, etc. |
| 4 | `RSD procedure type / level` | The type or level of Refugee Status Determination (RSD) procedure used to process the asylum claim. | Categorical | Different RSD procedure levels/types |
| 5 | `Tota pending start-year` | Number of asylum applications/cases that were still pending at the beginning of the year. | Numeric | Number of pending cases |
| 6 | `of which UNHCR-assisted(start-year)` | Number of cases pending at the beginning of the year that were assisted or handled with UNHCR involvement. | Numeric | Subset of the start-year pending cases |
| 7 | `Applied during year` | Number of new asylum applications submitted during the year. | Numeric | Number of applications |
| 8 | `decisions_recognized` | Number of asylum cases during the year in which refugee status was recognized. | Numeric | Recognized cases |
| 9 | `decisions_other` | Number of decisions made during the year that resulted in an outcome other than recognition or rejection. | Numeric | Other decisions |
| 10 | `Rejected` | Number of asylum applications rejected during the year. | Numeric | Rejected cases |
| 11 | `Otherwise closed` | Number of asylum cases closed during the year for reasons other than recognition or rejection. | Numeric | Withdrawn, discontinued, or otherwise closed cases |
| 12 | `Total decisions` | Total number of asylum cases for which a decision was made during the year. | Numeric | Usually combines recognized, rejected, other, and otherwise closed cases |
| 13 | `Total pending end-year` | Number of asylum cases that remained pending at the end of the year. | Numeric | Pending cases carried into the following year |
| 14 | `of which UNHCR-assisted(end-year)` | Number of cases pending at the end of the year that were assisted or handled with UNHCR involvement. | Numeric | Subset of the end-year pending cases |

# Data Cleaning

```python
df.head()
```

- In depth check for categorical columns

```python
cat_cols = df.select_dtypes(include= 'object').columns
cat_cols
```

```python
for col in cat_cols:
    print(col)
    print(df[col].nunique())
    print(df[col].unique())
    print('-' * 100)
```

```python
# Correct column name

df = df.rename(columns= {'Tota pending start-year' : 'Total pending start-year'})
df.head()
```

- convert datatype for columns (object --> float):

```python
df['Applied during year'] = df['Applied during year'].replace('*', np.nan)
df['Applied during year'] = pd.to_numeric(df['Applied during year'])
```

```python
num_cols = ['Total pending start-year','of which UNHCR-assisted(start-year)', 'Applied during year',	'decisions_recognized',
                'decisions_other','Rejected',	'Otherwise closed',	'Total decisions',	'Total pending end-year','of which UNHCR-assisted(end-year)']

for col in num_cols:
    df[col] = df[col].replace('*', np.nan)
    df[col] = pd.to_numeric(df[col])
```

```python
df.info()
```

- In depth check for Numerical columns

```python
num_cols = df.select_dtypes(include= 'number').columns
num_cols
```

```python
#for col in num_cols:

    #px.histogram(data_frame= df, x= col, title= col).show()
    #px.box(data_frame= df, x= col, title= col).show()
```

```python
# Drop Nan rows from columns ---> Low Nan %

df.dropna(subset= ['Total pending start-year', 'of which UNHCR-assisted(start-year)','Applied during year',
                  'decisions_recognized', 'Rejected','Otherwise closed','decisions_other',                           
                  'Total pending end-year','of which UNHCR-assisted(end-year)'], ignore_index= True, inplace= True)
```

```python
# Check duplicates after removing any column

df.duplicated().sum()
```

```python
# Check missing values by percentage

df.isna().mean().round(2) * 100
```

# Data Analysis

```python
df.head()
```

#### Univariate

### Q: How are asylum applications distributed across years?


```python
px.histogram(data_frame= df, x='Year' , y='Applied during year', text_auto= True, title= 'Total_Application_Year', nbins= 50)
```

```python
px.box(data_frame= df, x= 'Year')
```

## Q: What is the Percentage for Recognition & Rejection

```python
decision_outcomes = pd.DataFrame({'Outcome': ['Recognized', 'Rejected'],
                                 'Count': [df['decisions_recognized'].sum(),
                                  df['Rejected'].sum()]})
```

```python
px.pie(decision_outcomes,
    names='Outcome', values='Count',
    hole=0.5, title='Asylum Decision Outcomes')
```

#### Bivariate

###  Q: Does the number of applications increase or decrease over time?

```python
application_o_time = df.groupby('Year')[['Applied during year', 'Total decisions', 'Total pending end-year']].sum().reset_index()
application_o_time
```

```python
px.line(data_frame= application_o_time, x= 'Year', y= ['Applied during year', 'Total decisions', 'Total pending end-year'],
       title= 'Applications_Over_Time')
```

### Q: Which asylum/residence countries have the highest number of applications?
          (Top Asylum-Hosting Countries)

```python
highest_asylum_Hosting = (df.groupby('Country / territory of asylum/residence')['Applied during year']
                        .sum().sort_values(ascending=False).reset_index().head(10))
highest_asylum_Hosting
```

```python
px.pie(data_frame= highest_asylum_Hosting, names= 'Country / territory of asylum/residence', 
       values= 'Applied during year',
       title= 'Top Asylum-Hosting Countries',
       hole= 0.3)
```

### Q: Which origins are most frequently represented?

```python
top_origin = (df.groupby('Origin')['Applied during year']
        .sum()
        .sort_values(ascending=False)
        .head(10)
        .reset_index())
top_origin
```

```python
px.bar(data_frame=top_origin,
       x='Origin',
       y='Applied during year',
    title='Top 10 Countries of Origin')
```

### Q: Is there a relationship between applications and pending cases?

```python
df_corr = df[['Applied during year', 'Total pending end-year']].corr()
df_corr
```

```python
px.imshow(df_corr, text_auto= True)
```

### Multivariate

### Q: Does the number of decisions & rejecting keep up with applications?

```python
df_filtered =  df.groupby('Year')[['Applied during year', 
                                         'Total decisions','Total pending end-year',
                                           'decisions_recognized','Rejected']].sum().reset_index()
```

```python
px.bar(data_frame= df_filtered, x= 'Rejected',
       y='decisions_recognized', color= 'Year',
       title='Asylum Decisions by Outcome',
       barmode= 'group')
```

### Q: how asylum applications and outcomes changed over time for different countries of origin.

```python
origin_year = (df.groupby(['Year', 'Origin'])['Applied during year']
                .sum().head(10)
                .reset_index())

origin_year
```

```python
px.bar(data_frame= origin_year, x= 'Origin', y= 'Year',
       color= 'Applied during year',
       barmode= 'group', 
       text_auto= True)
```

### Q: Which countries have the largest pending asylum caseload?

```python
country_pending = (df.groupby('Country / territory of asylum/residence')[['Applied during year',
                    'Total decisions', 'Total pending end-year',
                      'decisions_recognized', 'Rejected']].sum().head(10)).sort_values('Total pending end-year', ascending=False).reset_index()
country_pending
```

```python
px.bar(data_frame= country_pending,
    x='Country / territory of asylum/residence',
    y='Total pending end-year',
    title='Total pending end-year')
```
